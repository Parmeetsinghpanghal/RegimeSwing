```python
# @title 2ï¸âƒ£ Initialize Paper Trading Engine
import yaml
import sqlite3
from datetime import datetime

with open('/content/RegimeSwing/config.yaml') as f:
    CONFIG = yaml.safe_load(f)

# State files on Google Drive (persistent)
STATE_DB = '/content/drive/MyDrive/RegimeSwing/state/paper_state.db'
TRADE_LOG = '/content/drive/MyDrive/RegimeSwing/paper_logs/paper_trades.csv'

# Initialize database
conn = sqlite3.connect(STATE_DB)
conn.execute("""
CREATE TABLE IF NOT EXISTS portfolio (
    key TEXT PRIMARY KEY,
    value TEXT,
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")
conn.execute("""
CREATE TABLE IF NOT EXISTS positions (
    symbol TEXT PRIMARY KEY,
    quantity INTEGER,
    avg_entry_price REAL,
    current_price REAL,
    unrealized_pnl REAL,
    realized_pnl REAL,
    entry_time TEXT,
    stop_price REAL,
    targets TEXT,
    strategy TEXT,
    delta REAL, gamma REAL, theta REAL, vega REAL,
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")
conn.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id TEXT PRIMARY KEY,
    symbol TEXT,
    side TEXT,
    order_type TEXT,
    quantity INTEGER,
    price REAL,
    trigger_price REAL,
    status TEXT,
    filled_quantity INTEGER,
    avg_fill_price REAL,
    timestamp TEXT,
    strategy TEXT,
    metadata TEXT
)
""")
conn.commit()
conn.close()

# Initialize CSV trade log
import pandas as pd
if not os.path.exists(TRADE_LOG):
    pd.DataFrame(columns=[
        'timestamp', 'symbol', 'side', 'quantity', 'price',
        'strategy', 'order_id', 'pnl', 'commission', 'slippage'
    ]).to_csv(TRADE_LOG, index=False)

print("âœ… Persistent state initialized")
```

```python
# @title 3ï¸âƒ£ Run Daily Paper Trading Cycle
from data.fetcher import create_fetcher
from indicators.technical import add_all_indicators, ensure_multiindex
from indicators.regime import RegimeDetector
from strategies.swing import run_all_strategies
from papertrade.engine import PaperTradingEngine, Order, OrderSide, OrderType, OrderStatus, Portfolio, Position
import uuid

fetcher = create_fetcher('/content/RegimeSwing/config.yaml')
engine = PaperTradingEngine(CONFIG, fetcher)
engine._load_state()  # Loads from SQLite on Drive

# Fetch current data
symbols = fetcher.nifty200_symbols[:50]
df = fetcher.get_price_data(symbols, period='3mo', interval='1d')
df = ensure_multiindex(df)
df = add_all_indicators(df, CONFIG)

# Regime
nifty_df = fetcher.get_nifty_index_data(period='3mo', interval='1d')
nifty_df = ensure_multiindex(nifty_df)
nifty_df = add_all_indicators(nifty_df, CONFIG)

detector = RegimeDetector(CONFIG)
regime = detector.detect_nifty_regime(nifty_df)
regime_info = {
    'regime': regime.regime,
    'size_mult': detector.get_size_multiplier(regime),
    'nifty_trend': regime.nifty_trend,
    'should_trade_swing': detector.should_trade_swing,
    'should_sell_options': detector.should_sell_options
}

# Generate signals
from strategies.swing import run_all_strategies
swing_signals = run_all_strategies(df, CONFIG, regime_info)

# Current prices
prices = {}
for sym in symbols:
    try:
        prices[sym] = df.xs(sym, level='Symbol')['Close'].iloc[-1]
    except:
        pass

# Run daily cycle
report = engine.run_daily_cycle(swing_signals, [], prices)

# Save state
engine._save_state()

print(f"\n=== DAILY REPORT: {datetime.now().strftime('%Y-%m-%d')} ===")
for k, v in report.items():
    if k != 'positions':
        print(f"  {k}: {v}")

print(f"\nðŸ“ˆ Positions ({len(report['positions'])}):")
for sym, pos in report['positions'].items():
    print(f"  {sym}: {pos['quantity']} @ {pos['avg_entry_price']:.2f} | "
          f"Current: {pos['current_price']:.2f} | PnL: â‚¹{pos['unrealized_pnl']:,.0f}")
```

```python
# @title ðŸ”Ÿ Risk Guard Check
from papertrade.engine import RiskGuards

risk_guards = RiskGuards(CONFIG)
risk_guards.reset_daily(engine.portfolio.total_value)

ok, msg = risk_guards.check_intraday(engine.portfolio)
print(f"Daily DD Check: {'âœ…' if ok else 'ðŸš¨'} {msg}")

ok, msg = risk_guards.check_options_greeks(engine.portfolio)
print(f"Greeks Check: {'âœ…' if ok else 'ðŸš¨'} {msg}")

# Portfolio Greeks
total_delta = sum(p.delta * p.quantity for p in engine.portfolio.positions.values())
total_theta = sum(p.theta * p.quantity for p in engine.portfolio.positions.values())
total_vega = sum(p.vega * p.quantity for p in engine.portfolio.positions.values())
print(f"\nPortfolio Greeks: Delta={total_delta:.2f}, Theta={total_theta:.2f}, Vega={total_vega:.2f}")
```

```python
# @title ðŸ”Ÿ Telegram Alert Setup (Optional)
# Set these in Colab secrets or environment variables:
# TELEGRAM_BOT_TOKEN, TELEGRAM_CHAT_ID

import os
import requests

BOT_TOKEN = os.environ.get('TELEGRAM_BOT_TOKEN', '')
CHAT_ID = os.environ.get('TELEGRAM_CHAT_ID', '')

def send_telegram(message: str):
    if not BOT_TOKEN or not CHAT_ID:
        return
    url = f"https://api.telegram.org/bot{BOT_TOKEN}/sendMessage"
    requests.post(url, json={'chat_id': CHAT_ID, 'text': message, 'parse_mode': 'HTML'})

# Send daily summary
if BOT_TOKEN and CHAT_ID:
    summary = f""
    <b>ðŸ“Š RegimeSwing Daily Report</b>
    Date: {datetime.now().strftime('%Y-%m-%d')}
    Regime: {regime.regime} ({regime.confidence:.0%})
    Portfolio: â‚¹{engine.portfolio.total_value:,.0f}
    Daily PnL: â‚¹{report['daily_pnl']:,.0f}
    Drawdown: {report['drawdown_pct']:.2%}
    Positions: {len(report['positions'])}
    """
    send_telegram(summary)
    print("âœ… Telegram alert sent")
else:
    print("â„¹ï¸ Telegram not configured (set TELEGRAM_BOT_TOKEN & TELEGRAM_CHAT_ID)")
```

## ðŸ“… Daily Routine

1. **Open notebook** â†’ **Runtime â†’ Run all** (2-3 min)
2. **Check output** for:
   - New signals generated
   - Stop/target hits
   - Risk guard alerts (ðŸš¨ = action needed)
   - Portfolio Greeks (Delta, Theta, Vega)
3. **Telegram alert** arrives automatically (if configured)
4. **State auto-saved** to Google Drive â€” survives restarts

## ðŸš¨ Action Items

| Alert | Action |
|-------|--------|
| Daily DD > 3% | Reduce position sizes |
| Daily DD > 5% | Stop new entries |
| Total DD > 12% | Reduce all positions 50% |
| Portfolio Delta > 0.3 | Hedge or reduce |
| Net Theta < 0 | Add premium selling |
| Correlation breach | Close overlapping positions |
