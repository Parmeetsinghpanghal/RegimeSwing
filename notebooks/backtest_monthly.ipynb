```python
# @title 3ï¸âƒ£ Backtest Configuration
import yaml

with open('/content/RegimeSwing/config.yaml') as f:
    CONFIG = yaml.safe_load(f)

BACKTEST_CONFIG = {
    'start_date': '2018-01-01',
    'end_date': '2024-12-31',
    'initial_capital': CONFIG['capital']['initial'],
    'max_risk_per_trade': CONFIG['capital']['max_risk_per_trade'],
    'max_positions': CONFIG['capital']['max_positions'],
    'commission_per_order': CONFIG['costs']['brokerage_per_order'],
    'slippage_bps': CONFIG['costs']['slippage_bps_equity'],
    'stt_rate': CONFIG['costs']['stt_equity_delivery'],
    'walk_forward_train_years': 3,
    'walk_forward_test_years': 1,
    'walk_forward_step_years': 1,
    'monte_carlo_sims': 500,
    'block_size': 20,
}

print("âœ… Config loaded")
print(f"Capital: â‚¹{BACKTEST_CONFIG['initial_capital']:,}")
print(f"Risk/trade: {BACKTEST_CONFIG['max_risk_per_trade']:.1%}")
```

```python
# @title 4ï¸âƒ£ Fetch Nifty 200 Data (Cached)
import pandas as pd
import numpy as np
from data.fetcher import create_fetcher
from indicators.technical import add_all_indicators, ensure_multiindex

fetcher = create_fetcher('/content/RegimeSwing/config.yaml')
symbols = fetcher.nifty200_symbols[:50]  # Top 50 for speed

print(f"Fetching {len(symbols)} symbols...")
df = fetcher.get_price_data(symbols, period='2500d', interval='1d')

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)
df.columns = [c.capitalize() for c in df.columns]

df = ensure_multiindex(df)
df = add_all_indicators(df, CONFIG)

# Nifty index for regime
nifty_df = fetcher.get_nifty_index_data(period='2500d', interval='1d')
if isinstance(nifty_df.columns, pd.MultiIndex):
    nifty_df.columns = nifty_df.columns.get_level_values(0)
nifty_df.columns = [c.capitalize() for c in nifty_df.columns]
nifty_df = ensure_multiindex(nifty_df)
nifty_df = add_all_indicators(nifty_df, CONFIG)

print(f"âœ… Data: {df.index.get_level_values('Symbol').nunique()} symbols, {len(df)} bars")
```

```python
# @title 5ï¸âƒ£ Run Walk-Forward Backtest
from backtest.engine import run_backtest, BacktestConfig
from indicators.regime import RegimeDetector
from strategies.swing import run_all_strategies

bt_config = BacktestConfig(**BACKTEST_CONFIG)

def strategy_func(df, _, regime_info):
    return run_all_strategies(df, CONFIG, regime_info)

detector = RegimeDetector(CONFIG)

print("ðŸ”„ Running walk-forward backtest...")
result = run_backtest(bt_config, fetcher, strategy_func, symbols[:30], detector)

print("\n=== BACKTEST RESULTS ===")
for k, v in result.metrics.items():
    if isinstance(v, float):
        print(f"  {k}: {v:.4f}")
    else:
        print(f"  {k}: {v}")
```

```python
# @title 6ï¸âƒ£ Monte Carlo Block Bootstrap
print("ðŸŽ² Running Monte Carlo ({} sims)...".format(BACKTEST_CONFIG['monte_carlo_sims']))

mc = result.monte_carlo
print(f"\n=== MONTE CARLO RESULTS ({len(mc.get('returns', []))} sims) ===")
for k, v in mc.items():
    if isinstance(v, float):
        print(f"  {k}: {v:.4f}")
    else:
        print(f"  {k}: {v}")
```

```python
# @title 7ï¸âƒ£ Regime-Attributed Performance
if not result.regime_performance.empty:
    print("\n=== REGIME PERFORMANCE ===")
    print(result.regime_performance.to_string())
else:
    print("Regime attribution not available")
```

```python
# @title 8ï¸âƒ£ Parameter Sensitivity Heatmaps
import plotly.express as px
import plotly.graph_objects as go

# Example: Sensitivity of trend_pullback to ADX threshold & RSI range
print("ðŸ“Š Generating sensitivity analysis...")

# This would run mini-backtests across parameter grid
# For demo, showing structure:
sensitivity_params = {
    'trend_pullback': {
        'adx_min': [20, 25, 30],
        'rsi_min': [35, 40, 45],
        'rsi_max': [50, 55, 60],
    },
    'breakout_volume': {
        'volume_mult': [1.2, 1.5, 2.0],
        'lookback_days': [15, 20, 25],
    }
}

print("\n=== SENSITIVITY GRIDS ===")
for strat, params in sensitivity_params.items():
    print(f"\n{strat}:")
    for p, values in params.items():
        print(f"  {p}: {values}")
```

```python
# @title 9ï¸âƒ£ Save Results to Google Drive
import json
from datetime import datetime

output = {
    'date': datetime.now().isoformat(),
    'config': BACKTEST_CONFIG,
    'metrics': {k: float(v) if isinstance(v, (int, float)) else str(v) 
              for k, v in result.metrics.items()},
    'monte_carlo': {k: float(v) if isinstance(v, (int, float)) else str(v) 
                    for k, v in result.monte_carlo.items()},
    'regime_performance': result.regime_performance.to_dict() 
                       if not result.regime_performance.empty else {},
}

filename = f"/content/drive/MyDrive/RegimeSwing/backtests/backtest_{datetime.now().strftime('%Y%m%d')}.json"
with open(filename, 'w') as f:
    json.dump(output, f, indent=2, default=str)

print(f"âœ… Saved to: {filename}")

# Also save equity curve
result.equity_curve.to_csv(filename.replace('.json', '_equity.csv'))
print("âœ… Equity curve saved")
```
